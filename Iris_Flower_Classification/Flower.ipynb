# Train-Test Split

I will be using pytorch for building the model, So we need to convert the data into tensors

```python
import torch
from sklearn.model_selection import train_test_split
```

```python
X = torch.tensor(x, dtype = torch.float32)
Y = torch.tensor(y, dtype = torch.int64)
```

```python
print(X)
print(Y)
```

```python

print(X.shape)
print(Y.shape)
```

```python
# I am splitting the data into temp train and test, which is 80% - 20%
X_train_temp, X_test, Y_train_temp, Y_test = train_test_split(X, Y, test_size= 0.2, random_state = 18, stratify = y.copy())
```

```python
# Now I will split the temp train into train and validation, which is 75% - 25% of temp train
X_train, X_val, Y_train, Y_val = train_test_split(X_train_temp, Y_train_temp, test_size = 0.25, random_state = 18, stratify = Y_train_temp)
```

# Neural Network Modelling

```python
import torch.nn as nn
```

```python
# Neural Network Class

class IrisClassifier(nn.Module):
    def __init__(self):
        super(IrisClassifier, self).__init__()
        self.layer1 = nn.Linear(4, 64) # input Layer (4 features) to hidden Layer (8 nodes)

        self. layer2 = nn.Linear(64, 3) # hidden Layer (8 nodes) to output Layer (3 classes)

    def forward(self, x):
        x = torch.relu(self.layer1(x)) # Activation function for the first hidden layer
        x = self.layer2(x) # No activation function for Output layer
        return x
```

```python
model = IrisClassifier()
print(model)
```

```python
# defining loss function
loss_function = nn.CrossEntropyLoss()

# defining optimizer
optimizer = torch.optim.Adam(model.parameters(), lr = 0.01)
```

# Evaluation and Prediction

```python
# creating the DataLoader for Batch Processing

from torch.utils.data import TensorDataset
from torch.utils.data import DataLoader

import seaborn as sns
```

```python
train_dataset = TensorDataset(X_train, Y_train)
val_dataset = TensorDataset(X_val, Y_val)
batch_size = 16
```

```python
train_loader = DataLoader(train_dataset, batch_size = batch_size, shuffle = True)
val_loader = DataLoader(val_dataset, batch_size = batch_size, shuffle = True)
```

### Training and Validation Loop

```python
num_epochs = 50
training_history = {'train_loss' : [], 'val_loss' : []}
print("Start Training....")

for epoch in range(num_epochs):
    # Training Phase
    model.train()
    run_loss = 0.0

    for inputs, labels in train_loader:
        optimizer.zero_grad() # set all the gradients to zero before calculating the new gradients
        
        output = model(inputs) # this is the forward pass
        loss = loss_function(output, labels) # calculating the loss
        loss.backward() # Backwardpropagation to calculate the gradients
        optimizer.step() # update the weights

        run_loss += loss.item() * inputs.size(0) # accumulate the loss over the batch
    
    epoch_loss = run_loss / len(train_dataset)
    training_history['train_loss'].append(epoch_loss)

    #validation Phase
    model.eval()
    correct = 0
    total = 0

    with torch.no_grad():
        for inputs, labels in val_loader:
            output = model(inputs)
            _, predication = torch.max(output.data, 1)
            total += labels.size(0)
            correct += (predication == labels).sum().item()

    val_accuracy = correct / total * 100
    training_history['val_loss'].append(val_accuracy)

    print(f"Epoch [{epoch} / {num_epochs}] , Train Loss: {epoch_loss:.4f} , Val Accuracy: {val_accuracy:.2f}%")

```

```python
# Now testing the model on the test dataset

model.eval()
with torch.no_grad():
    output = model(X_test)
    _, predicted_class = torch.max(output.data, 1)

# converting the tensor to numpy array
predicted_class = predicted_class.numpy()
print(predicted_class)
true_class = Y_test.numpy()
test_accuracy = np.sum(predicted_class == true_class) / len(true_class) * 100
print(f"Test Accuracy: {test_accuracy:.2f}%")
```

```python
from sklearn.metrics import confusion_matrix
```

```python
# plotting the training loss and validation accuracy
cm = confusion_matrix(true_class, predicted_class)
print(f"\nConfusion Matrix: {cm}")

plt.figure(figsize = (12, 5))
sns.heatmap(cm, annot = True, fmt = 'd', cmap = 'Blues', xticklabels = names, yticklabels = names)
plt.xlabel('Predicted')
plt.ylabel('True')
plt.title('Confusion Matrix')
plt.show()
```

Creating a Single prediction helper function (for flask to reuse)

```python
def predict_iris(sepal_length, sepal_width, petal_length, petal_width):
    model.eval()
    with torch.no_grad():
        x_single = torch.tensor([[sepal_length, sepal_width, petal_length, petal_width]], dtype = torch.float32)
        output = model(x_single)
        _, pred = torch.max(output.data, 1)
        label_idx = pred.item()
        return species_map[label_idx]
```

```python
print(predict_iris(5.1, 3.5, 1.4, 0.2))  # Expected output: 'setosa'
```
